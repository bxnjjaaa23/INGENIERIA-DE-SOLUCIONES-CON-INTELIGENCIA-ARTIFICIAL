## 2. Descarga Automática de Manuales y Archivos desde Google Drive Compartido
Haz click en el boton de ejecutar y se cargaran automaticamente los archivos necesarios


```python
# Descarga automática de manuales y archivos desde la carpeta pública de Google Drive
import gdown, os, shutil

DRIVE_PUBLIC_URL = 'https://drive.google.com/drive/folders/1UShpyjhW2rfinSafBseYJie8rw_P_dsK'
print('[Google Drive Público] Descargando archivos desde enlace compartido...')
try:
    gdown.download_folder(DRIVE_PUBLIC_URL, quiet=True)
    # Mover archivos al directorio raíz y eliminar la subcarpeta temporal para evitar duplicados
    folder_name = 'AutoTech_Manuales'
    if os.path.exists(folder_name):
        for fname in os.listdir(folder_name):
            src = os.path.join(folder_name, fname)
            dst = os.path.join('.', fname)
            if os.path.isfile(src):
                try:
                    shutil.move(src, dst)
                except Exception:
                    pass
        try:
            shutil.rmtree(folder_name)
        except Exception:
            pass
        print('[OK] Archivos movidos limpiamente a la raíz (sin duplicados).')
    else:
        print('[OK] Archivos descargados directamente en la raíz.')
except Exception as e:
    print(f'[Aviso] Error descargando con gdown: {e}')
```
